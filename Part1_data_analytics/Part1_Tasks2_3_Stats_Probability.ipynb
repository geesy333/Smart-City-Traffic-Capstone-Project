```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv('Metro_Interstate_Traffic_Volume.csv')

# ==========================================
# TASK 2: Descriptive Statistics & Correlation
# ==========================================
vol = df['traffic_volume']

mean_vol = vol.mean()
median_vol = vol.median()
std_vol = vol.std()
var_vol = vol.var()
range_vol = vol.max() - vol.min()
corr_temp_vol = df['temp'].corr(df['traffic_volume'])

print("=== TASK 2.1: TRAFFIC VOLUME STATISTICS ===")
print(f"Mean:               {mean_vol:,.2f}")
print(f"Median:             {median_vol:,.2f}")
print(f"Standard Deviation: {std_vol:,.2f}")
print(f"Variance:           {var_vol:,.2f}")
print(f"Range:              {range_vol:,} (Min: {vol.min()}, Max: {vol.max()})")

print("\n=== TASK 2.2: CORRELATION ANALYSIS ===")
print(f"Pearson Correlation (Temp vs. Traffic Volume): {corr_temp_vol:.4f}")

# ==========================================
# TASK 3: Probability & Congestion Analysis
# ==========================================
N = len(df)
congestion = df['traffic_volume'] > 5500
clear_weather = df['weather_main'] == 'Clear'
high_temp = df['temp'] > 292

# 3.1 Basic Probability
p_cong = congestion.mean()
p_clear = clear_weather.mean()
p_cong_and_clear = (congestion & clear_weather).mean()

print("\n=== TASK 3.1: BASIC PROBABILITY ===")
print(f"P(Congestion):                 {p_cong:.4f} ({p_cong*100:.2f}%)")
print(f"P(Clear Weather):              {p_clear:.4f} ({p_clear*100:.2f}%)")
print(f"P(Congestion AND Clear):       {p_cong_and_clear:.4f} ({p_cong_and_clear*100:.2f}%)")

# 3.2 Conditional Probability, Independence & Odds Ratio
p_clear_given_cong = (congestion & clear_weather).sum() / congestion.sum()
p_hightemp_given_cong = (congestion & high_temp).sum() / congestion.sum()
p_independent_expected = p_cong * p_clear

# Odds Ratio: Clear vs. Cloudy ('Clouds')
cloudy_weather = df['weather_main'] == 'Clouds'
p_cong_given_clear = (congestion & clear_weather).sum() / clear_weather.sum()
p_cong_given_cloudy = (congestion & cloudy_weather).sum() / cloudy_weather.sum()

odds_clear = p_cong_given_clear / (1 - p_cong_given_clear)
odds_cloudy = p_cong_given_cloudy / (1 - p_cong_given_cloudy)
odds_ratio = odds_clear / odds_cloudy

print("\n=== TASK 3.2: CONDITIONAL PROBABILITY & ODDS RATIO ===")
print(f"P(Clear Weather | Congestion):   {p_clear_given_cong:.4f} ({p_clear_given_cong*100:.2f}%)")
print(f"P(High Temp >292K | Congestion): {p_hightemp_given_cong:.4f} ({p_hightemp_given_cong*100:.2f}%)")
print(f"Independence Check: P(A ∩ B) = {p_cong_and_clear:.4f} vs P(A)*P(B) = {p_independent_expected:.4f}")
print(f"Odds of Congestion (Clear):      {odds_clear:.4f}")
print(f"Odds of Congestion (Clouds):     {odds_cloudy:.4f}")
print(f"Odds Ratio (Clear vs. Cloudy):   {odds_ratio:.4f}")
```
